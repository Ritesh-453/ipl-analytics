**Goal:** Explore team win records, toss decisions, top run scorers, and season-wise scoring trends.

**Key questions answered:**
- Which team has won the most matches overall?
- Does toss decision (bat/field) affect win rate?
- Who are the top 10 run scorers in IPL history?
- How has overall scoring evolved season by season?

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Count total wins for each team
wins = matches['winner'].value_counts().head(10)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=wins.values,
    y=wins.index,
    palette='Blues_r'
)

plt.title(
    'Top 10 Teams by Total Wins (2008-2024)',
    fontsize=14,
    fontweight='bold'
)

plt.xlabel('Number of Wins')
plt.ylabel('Team')

plt.tight_layout()

plt.savefig(
    '../../reports/figures/team_wins.png',
    dpi=150
)

plt.show()
```
